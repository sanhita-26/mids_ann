```python
import numpy as np
import matplotlib.pyplot as plt

def binary_step(x):
    return np.heaviside(x, 1)

x = np.linspace(-5, 5, 400)
y = binary_step(x)

plt.figure(figsize=(7,5))
plt.plot(x, y, 'black', linewidth=3)
plt.title("Binary Step", fontsize=18, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='k')
plt.axvline(0, color='k')
plt.ylim(-0.5, 1.5)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.linspace(-5, 5, 400)
y = sigmoid(x)

plt.figure(figsize=(7,5))
plt.plot(x, y, 'blue', linewidth=3, label='a')
plt.title("Sigmoid", fontsize=18, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='k')
plt.axvline(0, color='k')
plt.ylim(-1.2, 1.2)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def tanh(x):
    return np.tanh(x)

x = np.linspace(-5, 5, 400)
y = tanh(x)

plt.figure(figsize=(7,5))
plt.plot(x, y, 'red', linewidth=3)
plt.title("Tanh", fontsize=18, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='k')
plt.axvline(0, color='k')
plt.ylim(-1.2, 1.2)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def relu(x):
    return np.maximum(0, x)

x = np.linspace(-5, 5, 400)
y = relu(x)

plt.figure(figsize=(7,5))
plt.plot(x, y, 'green', linewidth=3)
plt.title("ReLU", fontsize=18, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='k')
plt.axvline(0, color='k')
plt.ylim(-2, 5)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def leaky_relu(x, alpha=0.1):
    return np.where(x > 0, x, x * alpha)

x = np.linspace(-5, 5, 400)
y = leaky_relu(x)

plt.figure(figsize=(7,5))
plt.plot(x, y, 'purple', linewidth=3)
plt.title("Leaky ReLU", fontsize=18, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='k')
plt.axvline(0, color='k')
plt.ylim(-2, 5)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def softplus(x):
    return np.log(1 + np.exp(x))

x = np.linspace(-5, 5, 400)
y = softplus(x)

plt.figure(figsize=(7,5))
plt.plot(x, y, 'orange', linewidth=3)
plt.title("Softplus", fontsize=18, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='k')
plt.axvline(0, color='k')
plt.ylim(-2, 5)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erf

def gelu(x):
    return 0.5 * x * (1 + erf(x / np.sqrt(2)))

x = np.linspace(-5, 5, 400)
y = gelu(x)

plt.figure(figsize=(7,5))
plt.plot(x, y, 'teal', linewidth=3)
plt.title("GELU", fontsize=18, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='k')
plt.axvline(0, color='k')
plt.ylim(-2, 5)
plt.show()
```

```python
from matplotlib import markers
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def swish(x):
    return x * sigmoid(x)

x = np.linspace(-5, 5, 400)
y = swish(x)

plt.figure(figsize=(7,5))
plt.plot(x, y, 'magenta', linewidth=3)
plt.title("Swish", fontsize=18, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='k')
plt.axvline(0, color='k')
plt.ylim(-2, 5)
plt.show()
```
