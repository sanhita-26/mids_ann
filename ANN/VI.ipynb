# 4B — Activation Functions Viva

---

## What is activation function?

Function deciding neuron output.

---

## Why activation functions needed?

Without activation functions:

* neural network becomes linear model only.

---

## What is sigmoid activation?

\sigma(x)=\frac{1}{1+e^{-x}}

Output range:

* 0 to 1.

---

## What is tanh activation?

tanh(x)=\frac{e^x-e^{-x}}{e^x+e^{-x}}

Output range:

* -1 to 1.

---

## What is ReLU?

ReLU(x)=max(0,x)

---

## Why ReLU widely used?

* fast computation
* reduces vanishing gradient problem.

---

## What is vanishing gradient problem?

Gradients become extremely small during deep learning training.

---

## Difference between sigmoid and ReLU?

| Sigmoid            | ReLU                       |
| ------------------ | -------------------------- |
| output 0–1         | output 0–∞                 |
| slower             | faster                     |
| vanishing gradient | reduced vanishing gradient |
