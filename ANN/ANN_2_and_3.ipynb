```python
import numpy as np

# Define network structure
input_neuron = 2
hidden_neuron = 4
output_neuron = 1
learning_rate = 0.5
epochs = 10000


# Random initialization
np.random.seed(42)


# XOR input dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Expected output
Y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Initialize weights
W1 = np.random.randn(input_neuron, hidden_neuron)
W2 = np.random.randn(hidden_neuron, output_neuron)

# Initialize biases
b1 = np.random.randn(1, hidden_neuron) * 0.01
b2 = np.random.randn(1, output_neuron) * 0.01


# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))


# Derivative of sigmoid
def sigmoid_derivative(x):
    return x * (1 - x)


# Training using backpropagation
for i in range(epochs):

    # Forward propagation
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    final_output = sigmoid(final_input)

    # Error calculation
    error = Y - final_output

    # Backpropagation
    output_delta = error * sigmoid_derivative(final_output)

    hidden_error = output_delta.dot(W2.T)

    hidden_delta = hidden_error * sigmoid_derivative(hidden_output)

    # Update weights and biases
    W2 += hidden_output.T.dot(output_delta) * learning_rate
    b2 += np.sum(output_delta, axis=0, keepdims=True) * learning_rate

    W1 += X.T.dot(hidden_delta) * learning_rate
    b1 += np.sum(hidden_delta, axis=0, keepdims=True) * learning_rate


# Testing
print("Predicted Output:\n")

print(final_output)

print("\nExpected Output:\n")

print(Y)
```
