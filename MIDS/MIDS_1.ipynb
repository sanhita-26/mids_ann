```python
df = pd.read_csv("Titanic-Dataset.csv")

df.head()
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
df.describe()
```

Replace Null Values with Mean for age and Embarked with mode

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())

print(df['Age'].isnull().sum())
```

```python
df['Embarked'] = df['Embarked'].fillna(
    df['Embarked'].mode()[0]
)

print(df['Embarked'].isnull().sum())
```

```python
df.drop('Cabin', axis=1, inplace=True)

print(df.columns)
```

```python
df.drop(
    ['PassengerId', 'Name', 'Ticket'],
    axis=1,
    inplace=True
)

df.head()
```

```python
# Encode Sex column
df['Sex'] = df['Sex'].map({
    'female': 0,
    'male': 1
})

# Encode Embarked column
df['Embarked'] = df['Embarked'].map({
    'S': 0,
    'C': 1,
    'Q': 2
})

df.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[['Age', 'Fare']] = scaler.fit_transform(
    df[['Age', 'Fare']]
)

df.head()
```

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
    
df.head()
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df.head()
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()
    [0])
print(df['Embarked'].isnull().sum())
df.head()
```

```python
df['FamilySize'] = (
    df['SibSp'] +
    df['Parch'] + 1
)

# Passenger alone or not
df['IsAlone'] = 1

df.loc[
    df['FamilySize'] > 1,
    'IsAlone'
] = 0
df.head()
```
